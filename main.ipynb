```python
from scipy.stats import zscore
import pandas as pd

df = pd.read_csv('amazon.csv')
df = df.dropna()
print(df.head())
```

```python
#Information generale sur le dataset

df_shape = df.shape
print("Nombre de lignes du dataset:", df_shape[0])
print("Nombre de colonnes:", df_shape[1])

df.dtypes
```

```python
#Changement de type dans les colonnes numeriques

df["discounted_price"] = (df["discounted_price"].str.replace("₹", "", regex=False).str.replace("," , "", regex=False).astype(float))

df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
#Discount

df["discount_percentage"] = (
    df["discount_percentage"].str.replace("%", "", regex=False).astype(float)
)
```

```python
#Ratings
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

df["rating_count"] = (
    df["rating_count"]
    .str.replace(",", "", regex=False)
    .astype(int)
)
```

```python
print(df[["actual_price","discounted_price","discount_percentage", "rating", "rating_count"]].head())
```

```python
#Verification des doublons 

print("Lignes repetees",df.duplicated().sum())
print("Reviews repetees", df['review_id'].duplicated().sum())
print("Produits repetees",df['product_id'].duplicated().sum())
```

```python
df[df['review_id'].duplicated(keep=False)].head(20)
```

```python
#Enlever les doublons
df = df.drop_duplicates(subset='review_id')
```

```python
print(df.duplicated().sum())
print(df['review_id'].duplicated().sum())
```

```python
#Verification des valeurs anormales
df['z_price'] = zscore(df['actual_price'])

outliers = df[abs(df['z_price']) > 3]
print(outliers[['actual_price', 'product_name']])

#On regarde que il y a des valeurs avec des prix hautes, ceci provoque un z_score> 3, mais sur les categories comme TV, etc... Alors les valeurs sont legitimes et on as pas besoin de les supprimer
```

```python
#Conversion des prix en euros

INR_TO_EUR = 0.011

df['actual_price_eu'] = df['actual_price'] * INR_TO_EUR
df['discounted_price_eu'] = df['discounted_price'] * INR_TO_EUR

df[['actual_price_eu', 'discounted_price_eu']].head()
```
